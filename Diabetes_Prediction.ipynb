```python
#importing the necessary packages
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("diabetes.csv")
df.head()
```

```python
#Exploring the Data Set
print("dataset shape:", df.shape)
print(df.columns)
df.info()
print(df.isnull().sum())
df.describe()
```

```python
#Separating the data set
x = df.drop('Outcome', axis=1)
print(x.head())
```

```python
y = df['Outcome']
print(y.head())
```

```python
print(y.value_counts())
```

```python
#Performing train-test split
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=0)

print("Training Features:", x_train.shape)
print("Testing Features:",x_test.shape)
print("Training Target:",y_train.shape)
print("Testing Target:",y_test.shape)
```

```python
#Performig pre processing
#Missing values
print(df.isnull().sum())
```

```python
#Identifying 0 values and converting to missing values
Columns_to_check = ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']
for col in Columns_to_check:
  df[col] = df[col].replace(0, np.nan)
  print(df[col].isnull().sum())
```

```python
#Replacing missing values with Median
from sklearn.impute import SimpleImputer
#copies of training and testing features
x_train_imputed = x_train.copy()
x_test_imputed = x_test.copy()
#creating median imputer
imputer = SimpleImputer(strategy='median')

```

```python
#fit and transfrom training data and testing data
x_train_imputed[Columns_to_check] = imputer.fit_transform(x_train_imputed[Columns_to_check])
x_test_imputed[Columns_to_check] = imputer.transform(x_test_imputed[Columns_to_check])

#Verifying result
print("Training missing values:")
print(x_train_imputed.isnull().sum())
print("\nTesting missing values:")
print(x_test_imputed.isnull().sum())
```

```python
#Scale/standardize data
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
x_train_scaled = pd.DataFrame( scaler.fit_transform(x_train_imputed), columns=x_train_imputed.columns, index=x_train_imputed.index)
print(x_train_scaled.head())
x_test_scaled = pd.DataFrame( scaler.transform(x_test_imputed), columns=x_test_imputed.columns, index=x_test_imputed.index)
print(x_test_scaled.head())
```

```python
# Machine Learning Model Building
#Importing scikit learn libraries

from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
#Logistic regression is used.
#Objective is to predict a patient is diabetic or not with 0/1 representations.
#Logistic regression uses sigmoid function to convert predictions to probabilities between 0 and 1.

model = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler()),
    ('classifier', LogisticRegression(max_iter=1000))
])
print(model)
model.fit(x_train_scaled, y_train)
```

```python
#Model Training

#Training and testing data
columns_to_check = ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']
x_train_imputed = x_train.copy()
x_test_imputed = x_test.copy()
x_train_imputed[columns_to_check] = imputer.fit_transform(x_train_imputed[columns_to_check])
x_test_imputed[columns_to_check] = imputer.transform(x_test_imputed[columns_to_check])
```

```python
#training the model to fit the selected model
model.fit(x_train_imputed, y_train)
print ("Model training completed successfully")
```

```python
#Generating Predictions on the trained data
y_train_pred = model.predict(x_train_imputed)
print ("Actual Values:")
print(y_train.head(10).values)
print ("Predicted Values:")
print(y_train_pred[:10])
```

```python
#Calculating the training accuracy
from sklearn.metrics import accuracy_score
train_accuracy = accuracy_score(y_train, y_train_pred)
print("Training Accuracy:", train_accuracy*100,"%")
```

```python
#Generating Predictions on the testing data
y_test_pred = model.predict(x_test_imputed)
print ("Actual Values:")
print(y_test.head(10).values)
print ("Predicted Values:")
print(y_test_pred[:10])
```

```python
#Calculating the testing accuracy
test_accuracy = accuracy_score(y_test, y_test_pred)
print("Testing Accuracy:", test_accuracy*100,"%")
```

```python
#Calculating the models performance accuracy
from sklearn.metrics import classification_report
print("Classification Report:")
print(classification_report(y_test, y_test_pred))
```

```python
#Generating confusion matrix
from sklearn.metrics import confusion_matrix
print("Confusion Matrix:")
print(confusion_matrix(y_test, y_test_pred))
```

```python
#Comparison of training vs testing performance
print ("Training Accuracy:", train_accuracy*100,"%")
print ("Testing Accuracy:", test_accuracy*100,"%")
accuracy_score_diff = train_accuracy - test_accuracy
print ("Accuracy Score Difference:", accuracy_score_diff*100,"%")
```
